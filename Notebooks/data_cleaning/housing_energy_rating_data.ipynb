# BER Research Tool (SEAI): Column Dictionary

**Dataset:** 569,058 BER assessments (2021–2025), 252 columns. One row = one BER certificate.

**Tags:**
- **[TARGET]**: what we want to predict
- **[LEAKAGE]**: calculated from the rating itself, so do **NOT** use as a model input
- **(?)**: meaning not 100% confirmed; check SEAI DEAP manual

---

## 1. Location, Time and Purpose
| Column | Meaning |
|---|---|
| CountyName | County of the home ("Co. Kildare"), or Dublin postal district ("Dublin 6") |
| DwellingTypeDescr | Type of home: detached, semi-detached, terraced, apartment, etc. |
| Year_of_Construction | Year the home was built |
| TypeofRating | Existing home, new home (final), or provisional (planned) rating |
| PurposeOfRating | Why the BER was done: sale, private letting, grant, new build, unknown |
| DateOfAssessment | Date the assessor did the BER |
| MultiDwellingMPRN | YES/NO: electricity meter (MPRN) shared by several homes |
| SA_Code | CSO Census Small Area code (very local area) |
| prob_smarea_error_0corr / prob_smarea_error_100corr | Estimated chance the Small Area code is wrong (?) |
| TGDLEdition | Edition of Building Regulations Part L (TGD L) the home was assessed against |

## 2. Rating Results (outputs of the BER calculation)
| Column | Meaning |
|---|---|
| BerRating | **[TARGET]** Energy use in kWh per m² per year (lower = better) |
| EnergyRating | **[TARGET]** BER band A1 (best) to G (worst), made from BerRating |
| CO2Rating | **[LEAKAGE]** CO₂ emissions in kg per m² per year |
| MPCDERValue | **[LEAKAGE]** Part L compliance CO₂/energy reference value (?) |
| EPC | **[LEAKAGE]** Energy Performance Coefficient (Part L check for new homes) |
| CPC | **[LEAKAGE]** Carbon Performance Coefficient (Part L check for new homes) |
| RER | **[LEAKAGE]** Renewable Energy Ratio (share of energy from renewables) |
| RenewEPnren / RenewEPren | **[LEAKAGE]** Non-renewable / renewable primary energy used for RER |
| HeatLossIndicator | **[LEAKAGE]** Heat loss per m² (used to judge if a heat pump is suitable) |
| GlobalWarmingPotentialValue | Global warming potential of refrigerant in heat pump/cooling (?) |
| SHMainPrimaryEnergy / SHSecPrimaryEnergy | **[LEAKAGE]** Primary energy for main / secondary space heating |
| WHMainPrimaryEnergy / WHSecPrimaryEnergy | **[LEAKAGE]** Primary energy for main / secondary water heating |
| PumpsFansPrimaryEnergy | **[LEAKAGE]** Primary energy for pumps and fans |
| LightingPrimaryEnergy | **[LEAKAGE]** Primary energy for lighting |
| CoolingPrimaryEnergy | **[LEAKAGE]** Primary energy for cooling |
| CHPInputPrimaryEnergy / CHPElecOutputPrimaryEnergy | **[LEAKAGE]** CHP energy in / electricity out |

|---|---|

### Key notes for modelling
1. **Targets:** `BerRating` (regression) and `EnergyRating` (classification).
2. **Never use [LEAKAGE] columns as inputs.** They come from the same calculation as the target.
3. `GroundFloorArea(sq m)` is the **total** floor area (it equals GroundFloorArea + FirstFloorArea + …).
4. Sections 8–12 are **mostly empty** (only a few homes have solar, CHP, wind etc.). Check % missing before using them.
5. `CountyName` mixes counties and Dublin postal districts. We need a county lookup to join with the PPR data.

```python
def stack_group(numbers, column_endings, new_names, number_name):
    """Turn columns like FirstWallArea, SecondWallArea... into rows."""
    pieces = []
    for n in numbers:
        # build the real column names, e.g. 'First' + 'WallArea' = 'FirstWallArea'
        old_cols = []
        for ending in column_endings:
            old_cols.append(n + ending)

        part = ber[['ber_id'] + old_cols].copy()
        part.columns = ['ber_id'] + new_names   # give simple names
        part[number_name] = n                   # remember which one it was (First, Second, ...)
        pieces.append(part)

    return pd.concat(pieces, ignore_index=True)


# ---------- floors ----------
floors = stack_group(['Ground', 'First', 'Second', 'Third'],
                     ['FloorArea', 'FloorHeight'],
                     ['floor_area', 'floor_height'], 'floor_no')
floors['floor_area'] = pd.to_numeric(floors['floor_area'], errors='coerce')
floors = floors[floors['floor_area'] > 0]          # keep only floors that exist

# ---------- walls ----------
walls = stack_group(['First', 'Second', 'Third'],
                    ['WallType_Description', 'WallDescription', 'WallArea', 'WallUValue',
                     'WallIsSemiExposed', 'WallAgeBandId', 'WallTypeId'],
                    ['wall_type', 'wall_note', 'wall_area', 'wall_uvalue',
                     'is_semi_exposed', 'age_band_id', 'wall_type_id'], 'wall_no')
walls['wall_area'] = pd.to_numeric(walls['wall_area'], errors='coerce')
walls = walls[walls['wall_area'] > 0]

# ---------- other energy sources ----------
other_energy = stack_group(['First', 'Second', 'Third'],
                           ['EnergyTypeId', 'EnergyType_Description', 'EnerProdComment',
                            'EnerProdDelivered', 'PartLTotalContribution', 'EnerProdConvFactor',
                            'EnerProdCO2EmissionFactor', 'EnerConsumedComment',
                            'EnerConsumedDelivered', 'EnerConsumedConvFactor',
                            'EnerConsumedCO2EmissionFactor'],
                           ['energy_type_id', 'energy_type', 'prod_comment', 'prod_delivered',
                            'partL_contribution', 'prod_conv_factor', 'prod_co2_factor',
                            'consumed_comment', 'consumed_delivered', 'consumed_conv_factor',
                            'consumed_co2_factor'], 'source_no')
other_energy['prod_delivered'] = pd.to_numeric(other_energy['prod_delivered'], errors='coerce')
other_energy['consumed_delivered'] = pd.to_numeric(other_energy['consumed_delivered'], errors='coerce')
other_energy = other_energy[(other_energy['prod_delivered'] > 0) | (other_energy['consumed_delivered'] > 0)]

# ---------- group heating boilers ----------
group_boilers = stack_group(['First', 'Second', 'Third'],
                            ['BoilerFuelType', 'HeatGenPlantEff', 'PercentageHeat'],
                            ['boiler_fuel', 'plant_efficiency', 'percentage_heat'], 'boiler_no')
group_boilers = group_boilers[group_boilers['boiler_fuel'].notna()]

# ---------- solar PV arrays ----------
solar_pv = stack_group(['First', 'Second', 'Third', 'Fourth', 'Fifth'],
                       ['SolarPVNoPanels', 'SolarPVPeakPowerPerPanel', 'SolarPVOrientation',
                        'SolarPVTilt', 'SolarPVOvershading'],
                       ['no_panels', 'peak_power_per_panel', 'orientation', 'tilt', 'overshading'],
                       'array_no')
solar_pv['no_panels'] = pd.to_numeric(solar_pv['no_panels'], errors='coerce')
solar_pv = solar_pv[solar_pv['no_panels'] > 0]

# ---------- wind turbines ----------
wind_turbines = stack_group(['First', 'Second', 'Third'],
                            ['WindTurbineRotorDiameter', 'WindTurbineHubHeight', 'WindTurbineTerrainType'],
                            ['rotor_diameter', 'hub_height', 'terrain_type'], 'turbine_no')
wind_turbines['rotor_diameter'] = pd.to_numeric(wind_turbines['rotor_diameter'], errors='coerce')
wind_turbines = wind_turbines[wind_turbines['rotor_diameter'] > 0]

print('One-to-many tables created')
```
