```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 200)   # show long text in full

file_path = r'C:\Users\LENOVO\OneDrive\OneDrive - National College of Ireland\Machine_learning_Project\dataSet\raw_dataset\Housing_dataset\Housing-plan\planning_applications_national.csv'
```

```python
planning = pd.read_csv(file_path, low_memory=False)

print(planning.shape)   # (rows, columns)
planning.head()
```

```python
print('Number of columns:', len(planning.columns))
planning.dtypes
```

## Step 4: Converting date columns
Dates are stored as form of large numbers, which is how ArcGIS saves them.
We convert them into normal dates type. easy to undersatnd and read

```python
date_columns = ['ReceivedDate', 'WithdrawnDate', 'DecisionDate', 'DecisionDueDate',
                'GrantDate', 'ExpiryDate', 'AppealDecisionDate', 'AppealSubmittedDate',
                'FIRequestDate', 'FIRecDate', 'ETL_DATE']

for col in date_columns:
    planning[col] = pd.to_datetime(planning[col], unit='ms', errors='coerce')

planning[['ApplicationNumber', 'ReceivedDate', 'DecisionDate', 'GrantDate']].head()
```

##Step 5: Check missing values (% of rows empty in each column)

```python
missing = planning.isnull().mean() * 100
missing.round(1).sort_values(ascending=False)
```

### Step 6: Look at the key columns
- Which councils are in the data?
- What decisions were made?
- What is the status of each application?
- What type of application is it?

```python
print(planning['PlanningAuthority'].value_counts(), '\n')
print(planning['Decision'].value_counts(dropna=False), '\n')
print(planning['ApplicationStatus'].value_counts(), '\n')
print(planning['ApplicationType'].value_counts())
```

## Step 8: Look at the text column (DevelopmentDescription)
The project brief needs text data averaging **at least 40 words per row**.

```python
# Show 5 example descriptions
for text in planning['DevelopmentDescription'].head(5):
    print('-', text)
    print()

# Count the words in each description
planning['word_count'] = planning['DevelopmentDescription'].fillna('').str.split().str.len()

print('Average words per description:', round(planning['word_count'].mean(), 1))
print('Median words per description: ', planning['word_count'].median())
```
